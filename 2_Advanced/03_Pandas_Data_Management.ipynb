```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

df_original = df.copy()  # Create a copy of the original DataFrame for backup
```

```python
#df.sample(5)
df.sample(10 ,random_state=42 )
df.sample(n=5)
df.sample(frac=0.1, replace=False)
```

```python
print('ID of df_original:               ', id(df_original))
print('ID of df_altered:                ', id(df_altered))
print('Are the two dataframes the same? ', id(df_original) == id(df_altered))
```
